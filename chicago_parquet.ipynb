```
# Cell 1: Import required libraries
from pyspark.sql import SparkSession
from pyspark.sql.types import *
from pyspark.sql.functions import *
import os
import uuid  # Added this import

# Cell 2: Configure storage account information
storage_account_name = "ayushdamg7370"
container_name = "silver"
tsv_file_path = "chicago_inspections_no_duplicates.tsv"
parquet_file_name = "Chicago_All_Years_Combined.parquet"

# Cell 3: Set up authentication for ADLS Gen2
# There are multiple ways to authenticate - use the one that applies to your environment

# OPTION 1: Using storage account key (replace with your actual key)
# This is the simplest approach but not recommended for production
spark.conf.set(f"fs.azure.account.key.{storage_account_name}.dfs.core.windows.net", 
               "xiKI4QF0gMqaJ7OZPoVraXUoU9muR4hjI3yIhpv7bVqJ0GanpCmhG423ZypkxZifOZYPzn4/EPpi+AStj4q6cw==")



# Cell 4: Define paths based on your authentication method
# If using direct access (Options 1 or 2)
tsv_full_path = f"abfss://{container_name}@{storage_account_name}.dfs.core.windows.net/{tsv_file_path}"
parquet_full_path = f"abfss://{container_name}@{storage_account_name}.dfs.core.windows.net/{parquet_file_name}"

# If using mounting (Option 3), use these paths instead:
# tsv_full_path = f"/mnt/{storage_account_name}-{container_name}/{tsv_file_path}"
# parquet_full_path = f"/mnt/{storage_account_name}-{container_name}/{parquet_output_path}"


# Read the TSV file
tsv_path = f"abfss://{container_name}@{storage_account_name}.dfs.core.windows.net/{tsv_file_path}"
df = spark.read.format("csv") \
    .option("header", "true") \
    .option("delimiter", "\t") \
    .option("inferSchema", "true") \
    .load(tsv_path)

# Create a temporary directory with a fixed name
temp_dir = "temp_parquet_dir"
temp_path = f"abfss://{container_name}@{storage_account_name}.dfs.core.windows.net/{temp_dir}"

# Delete the temp directory if it already exists
try:
    dbutils.fs.rm(temp_path, recurse=True)
    print(f"Deleted existing temporary directory: {temp_path}")
except:
    print(f"Temporary directory doesn't exist yet: {temp_path}")

# Write to a single file using coalesce
print("Writing data to temporary location...")
df.coalesce(1).write.mode("overwrite").parquet(temp_path)

# List files in the temp directory
temp_files = dbutils.fs.ls(temp_path)
print("Files in temporary directory:")
display(temp_files)

# Find the part file in the temporary directory using Python list comprehension
part_files = [f.path for f in temp_files if f.name.startswith("part-") and f.name.endswith(".parquet")]

if part_files:
    part_file = part_files[0]
    
    # Copy the part file to the final destination with the desired name
    final_path = f"abfss://{container_name}@{storage_account_name}.dfs.core.windows.net/{parquet_file_name}"
    print(f"Copying from {part_file} to {final_path}")
    dbutils.fs.cp(part_file, final_path)
    
    # Clean up the temporary directory
    dbutils.fs.rm(temp_path, recurse=True)
    
    print(f"Successfully created single Parquet file: {parquet_file_name}")
else:
    print("Error: No part files were found in the temporary directory")
    print("Files found were:")
    for file in temp_files:
        print(f" - {file.name}")
```
